```python
mtcars = pd.read_csv("mtcars.csv")
```

```python
mtcars.head()
```

```python
plt.figure(figsize=(8,3))
sns.scatterplot(x="wt", y="mpg",data=mtcars)
plt.title("Relationship between Weight and mileage")
plt.xlabel("Gear")
plt.ylabel("mileage")
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.boxplot(x="wt", y="mpg",data=mtcars)
plt.title("Relationship between Weight and mileage")
plt.xlabel("Gear")
plt.ylabel("mileage")
plt.show()
```

```python
plt.figure(figsize=(4,4))
sns.barplot(x="gear", y="mpg", data=mtcars)
plt.show()
```
