# Data Encoding

Nominal means: Categories with NO order/ranking.
 * Examples : Country names, Blood groups
 * Using in One-Hot Encoding

Ordinal means: Categories WITH meaningful order/ranking.
* Using in Ordinal Encoding
* Example : | Size   |
| ------ |
| Small  |
| Medium |
| Large  |

#### 3. One-Hot Encoding using Sklearn

```python
# Import function to split dataset into training and testing data
from sklearn.model_selection import train_test_split

# Import One Hot Encoder for converting categorical data into numerical form
from sklearn.preprocessing import OneHotEncoder

# Split dataset into:
# x -> input features
# y -> target/output column

# df.iloc[:,0:4]  -> selecting columns from index 0 to 3 as input features
# df.iloc[:,-1]   -> selecting last column as target variable

# test_size=0.2   -> 20% data used for testing
# random_state=2  -> ensures same random split every time

x_train, x_test, y_train, y_test = train_test_split(
    df.iloc[:,0:4],
    df.iloc[:,-1],
    test_size=0.2,
    random_state=2
)

# Create OneHotEncoder object

# drop="first"
# Removes first category to avoid Dummy Variable Trap
# Example:
# Fuel = Petrol, Diesel
# Only one column needed because second can be understood automatically

# sparse=False
# Returns normal NumPy array instead of sparse matrix

ohe = OneHotEncoder(
    drop="first",
    sparse_output=False
)

# Fit encoder on training data and transform categorical columns

# fit_transform():
# 1. Learns categories from training data
# 2. Converts categories into encoded numerical columns

# Encoding columns:
# owner and fuel

x_train_new = ohe.fit_transform(
    x_train[["owner", "fuel"]]
)

# Transform testing data using already learned categories

# IMPORTANT:
# We use only transform() on test data
# because model should learn ONLY from training data

x_test_new = ohe.transform(
    x_test[["fuel", "owner"]]
)

# Combine numerical columns with encoded categorical columns

# x_train[["brand","km_driven"]].values
# Selecting numerical/non-encoded columns

# np.hstack()
# Horizontally combines arrays column-wise

final_train_data = np.hstack((
    x_train[["brand", "km_driven"]].values,
    x_train_new
))

# Print final processed training data
print(final_train_data)
```

#### 4. OneHotEncoding with Top Categories

```python
# Count frequency of each unique brand
count = df["brand"].value_counts()

# Find number of unique brands
df["brand"].nunique()

# Set threshold value
# Brands having frequency <= 100 will be replaced
threshold = 100

# Select brands whose count is less than or equal to threshold
replace = count[count <= threshold].index

# Replace low-frequency brands with "other"
# Then apply One Hot Encoding using get_dummies()

encoded_data = pd.get_dummies(
    df["brand"].replace(replace, "other")
)

# Display encoded dataframe
print(encoded_data)
```

# Column Transformer

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("covid_toy.csv")

df.sample(5)
```

```python
df["city"].value_counts()

# gender -> One-Hot Encoding
# fever -> SimpleImputer
# cough -> ordinal Encoding 
# city -> OneHotEncoder
# has_covid -> LabelEncoder
```

```python
x_train, x_test, y_train, y_test = train_test_split(df.drop(columns=["has_covid"]), df['has_covid'], test_size=0.2)
x_train
```

```python
si = SimpleImputer()
x_train_fever = si.fit_transform(x_train[["fever"]])
x_test_fever = si.fit_transform(x_test[["fever"]])
```

```python
oe = OrdinalEncoder(categories=[["Mild","Strong"]])
x_train_cough = oe.fit_transform(x_train[["cough"]])
x_test_cough = oe.fit_transform(x_test[["cough"]])
x_train_cough.shape
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop="first", sparse_output=False)
x_train_city_gender = ohe.fit_transform(x_train[["city", "gender"]])
x_test_city_gender = ohe.fit_transform(x_test[["city", "gender"]])
```

```python
x_train_age = x_train.drop(columns=["gender", "city", "cough", "fever"]).values
x_test_age = x_test.drop(columns=["gender", "city", "cough", "fever"]).values
```

```python
x_train_transformed = np.concatenate((x_train_city_gender, x_train_age, x_train_fever, x_train_cough),axis=1)
x_test_transformed = np.concatenate((x_test_age,x_test_city_gender,x_test_cough, x_test_fever),axis=1)
```

#### Now using Column transformer

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(transformers=[
    ("tnf1",SimpleImputer(), ["fever"]),
    ("tnf2",OrdinalEncoder(categories=[["Mild", "Strong"]]),["cough"]),
    ("tnf3", OneHotEncoder(sparse_output=False, drop="first"),["gender", "city"])
],remainder="passthrough")
```
