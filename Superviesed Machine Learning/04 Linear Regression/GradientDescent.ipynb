```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
x,y = make_regression(n_samples=4, n_features=1, n_informative=1, n_targets=1, noise=80, random_state=13)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

lr.fit(x,y )
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
plt.scatter(x,y)
plt.plot(x, lr.predict(x),color="red")
```

```python
# let apply gradient descent assuming slope is constant m = 78.35
# and let assume the starting value for intercept b =0
y_pred = ((78.35*x) + 0 ).reshape(4)
```

```python
plt.scatter(x,y)
plt.plot(x, lr.predict(x),color="red",label="OLS")
plt.plot(x,y_pred, color= "#00a65a",label="b=0")
plt.legend()
plt.show()
```

```python
m = 78.35
b = 0

loss_scope = -2 * np.sum(y - m*x.ravel() - b)
loss_scope
```

```python
# let take learning rate = 0.1

lr = 0.1

step_size = loss_scope*lr
step_size
```

```python
# Calculating new intercept
b = b - step_size
```

```python
b
```

```python
y_pred1 = ((78.35 * x) + b).reshape(4)
plt.scatter(x,y)
# plt.plot(x, lr.predict(x),color="red",label="OLS")
plt.plot(x, y_pred1,color="#00a65a",label="b = {}".format(b))
plt.plot(x,y_pred, color= "#A3E4D7",label="b=0")
plt.legend()
plt.show()
```
