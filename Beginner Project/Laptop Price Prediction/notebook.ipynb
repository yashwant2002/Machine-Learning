```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("./laptop_data.csv")
df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
df.drop(columns="Unnamed: 0", inplace=True)
```

```python
df.head()
```

```python
df["Ram"]=df["Ram"].str.replace("GB","").astype("int32")
```

```python
df["Weight"] = df["Weight"].str.replace("kg","").astype("float64")
```

```python
df.head()
```

```python
sns.displot(df["Price"], kde=True, edgecolor="White")
```

```python
df["Company"].value_counts().plot(kind="bar")
```

```python
sns.barplot(data=df, x="Company", y="Price")
plt.xticks(rotation="vertical")
plt.show()
```

```python
df["TypeName"].value_counts().plot(kind="bar")
```

```python
sns.scatterplot(data=df, x="Inches", y="Price")
```

```python
df["ScreenResolution"].value_counts()
```

```python
df["Touchscreen"] = df["ScreenResolution"].apply(lambda x :1 if "Touchscreen" in x else 0)
```

```python
df.sample(5)
```

```python
sns.barplot(data=df, x="Touchscreen", y="Price")
```

```python
df["IPS"] = df["ScreenResolution"].apply(lambda x :1 if "IPS" in x else 0)
```

```python
df.head()
```

```python
df["IPS"].value_counts().plot(kind="bar")
```

```python
sns.barplot(data=df, x="IPS", y="Price")
```

```python
new = df["ScreenResolution"].str.split("x", n=1, expand=True)
```

```python
new
```

```python
df["X_res"] = new[0]
df["Y_res"] = new[1]
```

```python
df.head()
```

```python
df['X_res'] = df['X_res'].str.replace(',','').str.findall(r'(\d+\.?\d+)').apply(lambda x:x[0])
```

```python

df['X_res'] = df['X_res'].astype('int')
df['Y_res'] = df['Y_res'].astype('int')
```
